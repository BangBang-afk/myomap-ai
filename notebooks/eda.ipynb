# MyoMap AI — EDA (local + Kaggle)
Kaggle input: `/kaggle/input/rsna-knee-abnormality-detection` auto-mounted. Local: `../data/raw`

```python
import os, glob
from pathlib import Path
import pandas as pd, numpy as np
KAGGLE = Path("/kaggle/input/rsna-knee-abnormality-detection")
LOCAL = Path("../data/raw")
ROOT = KAGGLE if KAGGLE.exists() else LOCAL
print(f"ROOT={ROOT} exists={ROOT.exists()}")
for f in sorted(ROOT.rglob("*.csv"))[:20]:
    print(f)
```

```python
if ROOT.exists():
    train = pd.read_csv(next(ROOT.rglob("train.csv"))) if list(ROOT.rglob("train.csv")) else None
    print(train.head() if train is not None else "no train.csv")
    if train is not None:
        label_cols = [c for c in train.columns if c not in ["StudyInstanceUID","report_text","report","fold"]]
        print("labels", label_cols[:12])
        print(train[label_cols].mean().sort_values())
        print(train[label_cols].isna().sum())
else:
    print("Mock mode — no data downloaded yet. Run: kaggle competitions download -c rsna-knee-abnormality-detection -p ../data/raw --unzip")
```

```python
# DICOM quick view
try:
    import pydicom, matplotlib.pyplot as plt
    dcm = next(ROOT.rglob("*.dcm"), None)
    if dcm:
        ds = pydicom.dcmread(str(dcm))
        print(ds)
        plt.imshow(ds.pixel_array, cmap="gray"); plt.show()
    else:
        print("no dcm found — mock")
except Exception as e:
    print(e)
```
